```python
from collections import Counter
```

```python
counter = Counter()
word_counter = Counter()
for line in open("../../deep_disfluency/deep_disfluency/data/lm_corpora/swbd_disf_train_1_2_clean.text"):
    spl = line.split(",")
    #print spl
    if spl[0] == "POS":
        continue
    words = spl[-1].lower().split()
    for word in words:
        word_counter[word] += 1
    letters = []
    for word in set(words):
        #print word
        letters = set(word)
        #print letters
        for l in list(letters):
            counter[l]+=1
        #print counter
        #break
```
